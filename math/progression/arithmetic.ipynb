```python
from IPython.display import display, Math
import sympy as sp
```

### Sum

```python
S, a1, a_last, n = sp.symbols('S, a1, a_last, n')

sum_exp = (a1 + a_last) / 2 * n - S

display(Math(sp.latex(sum_exp)))
```

```python
first = 1
last = 10
numbers = [*range(first, last), last]

result = sum_exp.subs({ a1: first, a_last: last, n: len(numbers) })

display(sp.solve(result, S)[0])
```
