### By loops

```python
def generate_fibonacci_sequence(n: int = 0) -> list[int]:
    sequence = []
    a, b = 0, 1
    
    for _ in range(n):
        prev_number = a
        sequence.append(prev_number)
        a, b = b, a + b
        
    return sequence
```

```python
generate_fibonacci_sequence(10)
```

### By recursion

```python
def generate_fibonacci_sequence_recursively(n: int = 0) -> list[int]:
    if n > 2:
        sequence = generate_fibonacci_sequence_recursively(n - 1)
        next_number = sequence[-2] + sequence[-1]
        sequence.append(next_number)
        return sequence
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        return []
```

```python
generate_fibonacci_sequence_recursively(10)
```
