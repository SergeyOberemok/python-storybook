## Fractions

### NOK - найбольшее общее кратное

#### by multiplication

```python
def get_nok_by_cross_multiplication(num1: int, num2: int) -> int:
    if num1 == num2:
        return num1
    
    return num1 * num2
```

```python
x = 6
y = 8

print('Common multiple (not necessarily the least) of %g and %g is %g' %(x, y, get_nok_by_cross_multiplication(x, y)))
```

#### by self division

```python
def get_nok_by_self_division(num1: int, num2: int) -> int | None:
    if num1 % num2 == 0:
        return num1
    
    if num2 % num1 == 0:
        return num2
    
    return None
```

```python
x = 60
y = 15

print('NOK of %g and %g is %g' %(x, y, get_nok_by_self_division(x, y)))
```

#### by sequence

```python
def get_nok_by_sequence(num1: int, num2: int) -> int:
    max_nok = get_nok_by_cross_multiplication(num1, num2)
    i = 1
    num1_dictionary = { num1: i }
    num2_dictionary = { num2: i }
    
    while num1 < max_nok and num2 < max_nok:
        last_key = list(num2_dictionary)[-1]
        if num1_dictionary.get(last_key) is not None:
            return last_key

        last_key = list(num1_dictionary)[-1]
        if num2_dictionary.get(last_key) is not None:
            return last_key
        
        i += 1
        
        num1_dictionary.update({ num1 * i: i })
        num2_dictionary.update({ num2 * i: i })
    
    return max_nok
```

```python
x = 6
y = 8

print('NOK of %g and %g is %g' %(x, y, get_nok_by_sequence(x, y)))
```

### NOD - найбольший общий делитель | GCD - greatest common denominator

#### number's multipliers

```python
def get_multipliers(num: int) -> list:
    multipliers = []
    
    for i in range(2, num + 1):
        while num % i == 0:
            multipliers.append(i)
            num //= i
    
    return multipliers
```

```python
x = 36
multipliers = get_multipliers(x);

print('Multipliers of %g is %s' %(x, ', '.join(map(str, multipliers))))
```

```python
factorints = sp.factorint(x);
multipliers = []

for key in factorints:
    list_of_multipliers = [key] * factorints[key]
    multipliers.extend(list_of_multipliers)
    
print('Multipliers of %g is %s' %(x, ', '.join(map(str, multipliers))))
```

#### multipliers intersections

```python
def get_intersections(list1, list2) -> list:
    list1 = list1.copy()
    list2 = list2.copy()
    intercestions = []
    
    for i in list1:
        if i in list2:
            intercestions.append(i)
            list2.remove(i)
        
    return intercestions
```

```python
x = [2, 2, 2, 2, 3]
y = [2, 2, 3, 3]

intersections = get_intersections(x, y)

print('Intersections of %s and %s are %s' %(
    ', '.join(map(str, x)), 
    ', '.join(map(str, y)), 
    ', '.join(map(str, intersections))
))
```

#### by multiplying

```python
def get_nod_by_multiplying(num1: int, num2: int) -> int:
    num1_multipliers = get_multipliers(num1)
    num2_multipliers = get_multipliers(num2)
    intersections = get_intersections(num1_multipliers, num2_multipliers)
    
    return functools.reduce(lambda a, b: a * b, intersections, 1)
```

```python
x = 48
y = 36

print('NOD | GCD between %g and %g is %g' %(x, y, get_nod_by_multiplying(x, y)))
print('NOD | GCD between %g and %g is %g' %(x, y, math.gcd(x, y)))
```

### Prime numbers

```python
def is_prime(num: int) -> bool:
    if num <= 3:
        return num > 1
    if num % 2 == 0 or num % 3 == 0:
        return False
    
    for i in range(5, int(num ** 0.5) + 1, 6):
        if num % i == 0 or num % (i + 2) == 0:
            return False
    return True
```

```python
x = 7

print('Is %g is prime = %s' %(x, 'yes' if is_prime(x) else 'no'))
```

#### sequence

```python
def get_prime_numbers(length: int = 0) -> list:
    if length <= 0:
        return []
    
    prime_numbers = []
    
    for i in range(1, length):
        if is_prime(i):
            prime_numbers.append(i)
    
    return prime_numbers
```

```python
x = 20
prime_numbers = get_prime_numbers(x)

print('Prime numbers till %g are %s' %(x, ', '.join(map(str, prime_numbers))))
```
