### Vars

```python
columns, gutter, margin, whole, x = sp.symbols("cols, gutter, margin, w, x")
```

### Grid

##### Expression

```python
grid_exp = columns * x + (columns - 1) * gutter - whole
```

```python
display(Math(sp.latex(grid_exp)))
```

##### Substituted

```python
grid_exp_substituted = grid_exp.subs(
    {
        columns: columns_widget.value,
        gutter: gutters_width_widget.value,
        whole: whole_width_widget.value - (margin_width * 2),
    }
)
```

```python
grid_exp_substituted
```

##### Result - Column width

```python
x_result = sp.solve(grid_exp_substituted.subs(margin, margin_width), x)
```

```python
next(iter(x_result))
```

### Middle line between

##### Expression

```python
middle_line_between_exp = columns * x + gutter * (columns - 1) + gutter / 2
```

```python
display(Math(sp.latex(middle_line_between_exp)))
```

##### Substituted

```python
middle_line_between_exp_substituted = middle_line_between_exp.subs({
    columns: columns_x,
    gutter: gutters_width_widget.value,
    x: x_result[0]
})
```

```python
middle_line_between_exp_substituted
```

### Plot

```python
w = whole_width_widget.value - (margin_width * 2)
golden_ration = calc_golden_ration_x(w)

plt.plot([0, w], [0, w - golden_ration])
plt.plot([0, w], [0, middle_line_between_exp_substituted])

plt.grid()
plt.show()

golden_ration, w - golden_ration, middle_line_between_exp_substituted
```
