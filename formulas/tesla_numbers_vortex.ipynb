```python
from IPython.display import display, Math
from decimal import Decimal
```

```python
# Digital root

def get_digital_root(num):
    if (num < 0): return 0
    
    # Decimal avoids float noise (0.1 * 10 != 1.0) and scientific notation (6.1e-05) in str(num)
    digits = format(Decimal(repr(num)), 'f')
    digits_sum = sum(int(digit) for digit in digits if digit.isdigit())
    
    if digits_sum >= 10:
        digits_sum = get_digital_root(digits_sum)

    return digits_sum

get_digital_root(128), get_digital_root(0.0125), get_digital_root(1.5)
```

```python
def get_range(length):
    if length >= 0:
        return range(1, length + 1)
    else:
        return reversed(range(length, 0))
```

```python
def get_multiplied(base_number, to_extend):
    if to_extend <= 0:
        return 0
    
    if to_extend == 1:
        return base_number
    
    return get_multiplied(base_number, to_extend - 1) * 2

get_multiplied(1, 12)
```

```python
def get_divided(base_number, to_extend):
    if to_extend <= 0:
        return 0
    
    if to_extend == 1:
        return base_number
    
    return get_divided(base_number, to_extend - 1) / 2

get_divided(1, 4)
```

```python
# Double digits

base_number = 1

for x in get_range(12):
    summed_number = get_multiplied(base_number, x)
    display(Math('%g = %g \\rightarrow %g' %(x, summed_number, get_digital_root(summed_number))))
```

```python
# Divide digits

base_number = 3

for x in get_range(-12):
    divided_number = get_divided(base_number, abs(x))
    display(Math('%g = %f \\rightarrow %g' %(x, divided_number, get_digital_root(divided_number))))
```

```python
# Digital root sequence

def get_digital_root_sequence(base_number, length):
    digital_roots = []
    
    for x in get_range(length):
        num = get_multiplied(base_number, x) if x > 0 else get_divided(base_number, abs(x)) 
        digital_roots.append(get_digital_root(num))
    
    return digital_roots

length = int(input('Enter length '))

print(get_digital_root_sequence(1, length))
print(get_digital_root_sequence(1, -length))
print(get_digital_root_sequence(3, length))
print(get_digital_root_sequence(3, -length))
```
