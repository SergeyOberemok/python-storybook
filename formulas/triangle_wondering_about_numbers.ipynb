#### Input triangle vertexes

```python
triangle_vertexes_number = 3
answers = []

for i in range(0, triangle_vertexes_number):
    answer = int(input('Enter number %g - ' %(i + 1)))
    
    answers.append(answer)

answers
```

#### Sum between adjacent numbers

```python
sums_of_adjacents = []

for i, answer in enumerate(answers):
    next_index = i + 1
    
    if next_index >= len(answers):
        next_index = 0
    
    sum_result = answer + answers[next_index]
    
    print('%g + %g = %g' %(answer, answers[next_index], sum_result))
    
    sums_of_adjacents.append(sum_result)

sums_of_adjacents
```

#### Sum of medians

```python
sums_of_medians = []

for i, answer in enumerate(answers):
    median_index = i + 1
    
    if median_index >= len(answers):
        median_index = 0
        
    sum_result = answer + sums_of_adjacents[median_index]
    
    print('%g + %g = %g' %(answer, sums_of_adjacents[median_index], sum_result))
    
    sums_of_medians.append(sum_result)
    
sums_of_medians
```
