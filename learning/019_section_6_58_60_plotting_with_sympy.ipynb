```python
import sympy as sp
```

### Plotting

```python
from sympy.abc import x

y = x**2

plot = sp.plotting.plot(y)
```

### Amend plot

```python
from sympy.abc import x, y

exp = x/y

y_limits = (-5, 5)

plot = sp.plotting.plot(exp.subs(x, 1), (y,) + y_limits, show = False)

plot.ylim = list(y_limits)
plot.show()
```

### Extend plot

```python
plot_to_extend = sp.plotting.plot(exp.subs(x, 2), show = False)
plot.extend(plot_to_extend)

plot.show()
```

### Excercise

```python
a = sp.symbols('a')

y = a/(x**2 - a)

plot = None
x_limits = (-5, 5)
y_limits = (-10, 10)

for i in range(1, 5):
    i_plot = sp.plotting.plot(y.subs(a, i), (x,) + x_limits, show = False)
    i_plot[0].label = '$%s$' %sp.latex(y.subs(a, i))
    
    if plot is None:
        plot = i_plot
    else:
        plot.extend(i_plot)

plot.ylim = list(y_limits)
plot.legend = True
plot.show()
```
