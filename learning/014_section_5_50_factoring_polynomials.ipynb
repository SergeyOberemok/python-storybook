```python
import sympy as sp
from IPython.display import display, Math
from sympy.abc import x
```

### Factor

#### Factorable

```python
factorable_exp = x**2 + 4*x + 3

display(Math(sp.latex(sp.Eq(factorable_exp, sp.factor(factorable_exp)))))
```

#### Not factorable

```python
not_factorable_exp = x**2 + 4*x - 3

display(Math(sp.latex(sp.Eq(not_factorable_exp, sp.factor(not_factorable_exp)))))
display(Math(sp.latex(sp.factor(not_factorable_exp))))
```
