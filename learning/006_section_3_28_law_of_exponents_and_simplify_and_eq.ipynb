```python
import sympy as sp
from IPython.display import display, Math
```

```python
x, y, z = sp.symbols('x, y, z')
```

```python
def display_equation(exp_left, exp_right):
    latex_exp_left = sp.latex(exp_left);
    latex_exp_right = sp.latex(exp_right)

    display(Math('%s = %s' %(latex_exp_left, latex_exp_right)))
```

#### Multiplication

```python
expression_left = multiplication_exp = x ** y * x ** z
expression_right = x ** (y + z)

display_equation(expression_left, expression_right)
```

#### Division

```python
expression_left = division_exp = x ** y / x ** z
expression_right = x ** (y - z)

display(Math('\\frac{%s^%s}{%s^%s} = %s' %(x, y, x, z, sp.latex(expression_right))))
```

#### Simplify

```python
display(Math(sp.latex(multiplication_exp)), sp.simplify(multiplication_exp))
display(Math(sp.latex(division_exp)), sp.simplify(division_exp))
```

#### Eq

```python
sp.Eq(4, 2 + 2)
```

```python
sp.Eq(multiplication_exp, sp.simplify(multiplication_exp))
```

```python
sp.Eq(division_exp, sp.simplify(division_exp))
```
