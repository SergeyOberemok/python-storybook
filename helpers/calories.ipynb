## Calories calculator

---

```python
from IPython.display import display, Math
import sympy as sp
from enum import Enum
```

---

## Calories quantity

#### Expresion

```python
def get_calories_exp():
    quantity, calories = sp.symbols('quantity, calories')

    calories_exp = quantity / 100 * calories
    
    return calories_exp
```

```python
calories_exp = get_calories_exp()

display(Math(sp.latex(calories_exp)))
```

#### Resolve

```python
def get_calories(quantity, calories):
    calories_exp = get_calories_exp()
    
    result = calories_exp.subs({ 'quantity': quantity, 'calories': calories }).evalf()
    
    return round(result, 2)
```

```python
quantity = 130
calories = 418

get_calories(quantity, calories)
```

## Water

#### Expresion

```python
net, water = sp.symbols('net, water')
total = net + water

water_exp = 100 - (net * 100 / total)

display(Math(sp.latex(water_exp)))
```

#### Resolve

```python
gross_weight = 2734
net_weight = 2246
water_weight = gross_weight - net_weight

result = round(water_exp.subs({ 'net': net_weight, 'water': water_weight }).evalf())
    
print(result)
```

## Calories equations

```python
weight, height, age, activity = sp.symbols('weight, height, age, activity')

class Activity(Enum):
    SEATED = 1.2
    LIGHT = 1.375
    MODERATE = 1.55
    ACTIVE = 1.725
    VERY_ACTIVE = 1.9
```

#### Harris-Benedict equation for men (kg, cm, year)

```python
harris_benedict_exp = (66.473 + (12.7516 * weight) + (5.0033 * height) - (6.755 * age)) * activity

display(Math(sp.latex(harris_benedict_exp)))
```

```python
params = { 
    weight: 80, 
    height: 180, 
    age: 27, 
    activity: Activity.SEATED.value 
}

bmr = harris_benedict_exp.subs(params).evalf()
round(bmr)
```

##### Input

```python
weight_input = float(input('Enter weight '))
height_input = float(input('Enter height '))
age_input = float(input('Enter age '))

bmr = harris_benedict_exp.subs({ weight: weight_input, height: height_input, age: age_input, activity: Activity.SEATED.value }).evalf()

print('Daily calories intakes is %s' % round(bmr))
```

#### Mifflin-St-Jeor equation for men (kg, cm, year)

```python
mifflin_st_jeor_exp = ((10 * weight) + (6.25 * height) - (5 * age) + 5) * activity

display(Math(sp.latex(mifflin_st_jeor_exp)))
```

```python
params = { weight: 80, height: 180, age: 27, activity: Activity.SEATED.value }

bmr = mifflin_st_jeor_exp.subs(params).evalf()
round(bmr)
```

##### Input

```python
weight_input = float(input('Enter weight '))
height_input = float(input('Enter height '))
age_input = float(input('Enter age '))

bmr = mifflin_st_jeor_exp.subs({ weight: weight_input, height: height_input, age: age_input, activity: Activity.SEATED.value }).evalf()

print('Daily calories intakes is %s' % round(bmr))
```
