```python
import os
from pathlib import Path
```

##### with os

```python
[f for f in os.listdir('./') if os.path.isfile(os.path.join('./', f)) and f.endswith('.txt')]
```

##### with pathlib

```python
dir_path = Path('./')

[p for p in dir_path.iterdir() if p.is_file() and p.suffix == '.txt']
```
