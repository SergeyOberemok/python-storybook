```python
from difflib import Differ, SequenceMatcher, unified_diff, HtmlDiff
from faker import Faker
from IPython.display import display, HTML
```

```python
faker = Faker()
```

---

```python
text = faker.text()
```

```python
print(text)
```

##### Ratio

```python
sequence_matcher = SequenceMatcher(None, text, text.replace('a', 'b'))

sequence_matcher.ratio()
```

##### Unified (Git style)

```python
diff = unified_diff(text, text.replace('a', 'b'), lineterm = '')

print(''.join(list(diff)))
```

##### Compare

```python
differ = Differ()

diff = differ.compare(text.splitlines(), text.replace('a', 'b').splitlines())

print(''.join(diff))
```

##### HTML

```python
html_differ = HtmlDiff()

html = html_differ.make_table(text.splitlines(), text.replace('a', 'b').splitlines())

display(HTML(html))
```
