```python
import pandas as pd
from faker import Faker
```

```python
fake = Faker()
```

---

```python
name_age_df = pd.DataFrame({
    'name': ['John', 'Mark'] + ['Barbie'],
    'age': [fake.random_number(digits=2) for _ in range(3)]
})

name_phone_df = pd.DataFrame({
    'name': ['Mark', 'John'] + ['Ken'],
    'phone': [fake.phone_number() for _ in range(3)]
})
```

### Join

```python
name_age_df.join(name_phone_df.set_index('name'), on='name', how='left').head() # set_index to prevent error: You are trying to merge on str and int64 columns for key 'name'
```

### merge

##### inner

```python
merged_df= name_age_df.merge(name_phone_df, on='name').head()
```

##### left

```python
name_phone_df['phone'] = name_phone_df['phone'].transform(lambda phone: fake.phone_number())
```

```python
merged_df.merge(name_phone_df, on='name', how='left', suffixes=['_old', '_new']).head()
```

##### right

```python
name_age_df['age'] = name_age_df['age'].transform(lambda age: age + 1)
```

```python
merged_df.merge(name_age_df, on='name', how='right', suffixes=['_old', '_new']).head()
```
