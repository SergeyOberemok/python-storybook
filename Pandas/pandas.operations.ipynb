```python
import pandas as pd
import numpy as np
from faker import Faker
```

```python
fake = Faker()
records_count = 10
```

```python
sexes = list(map(lambda _: 'M' if fake.boolean() else 'F', [*range(records_count)]))

df = pd.DataFrame({
    'name': list(map(lambda sex: fake.first_name_male() if sex == 'M' else fake.first_name_female(), sexes)),
    'sex': sexes,
    'children': list(map(lambda _: fake.random_digit(), [0] * records_count)),
    'income': list(map(lambda _: fake.random_int(min= 3000, max= 10000), [*range(records_count)])),
})

df.head()
```

```python
df.describe(percentiles=None, include='all')
```

### Mean

```python
df[['children']].mean()
```

### Median

```python
df[['income']].median()
```

### Mode

```python
df[['sex']].mode()
```

### Variance

```python
# Series and DataFrame normalizes by N-1 (unbiased variance), while numpy.var normalizes by N (biased variance)

df[['income']].mean(), df[['income']].var(), df['income'].var(), np.var(df['income'])
```

### Standard deviation

```python
df[['income']].mean(), df[['income']].std()
```
