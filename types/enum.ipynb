```python
from enum import Enum
import json
```

---

```python
class Person(str, Enum): # should be inherited from 'str' for json dumps
    MALE = 'male'
    FEMALE = 'female'
```

##### by property

```python
Person.MALE.name
```

```python
Person.FEMALE.value
```

##### by key

```python
assert Person.MALE == Person['MALE']

Person['MALE']
```

##### by value

```python
assert Person.FEMALE == Person('female')

Person('female')
```

```python
json.dumps(Person.MALE)
```

###### should be inherited from str in order dumps to work
