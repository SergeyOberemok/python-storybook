##### Deep down the tree

```python
def find_all_files(root_path: str, extension: str, exclude_dirs: tuple[str, ...] = ()):
    for root, dirs, files in os.walk(root_path):
        if any(excluded in root for excluded in exclude_dirs):
            continue

        for file in files:
            if file.lower().endswith(f'.{extension.lower()}'):
                yield os.path.normpath(os.path.join(root, file))
```
