# Definition

## Simple

```python
def sum_two_numbers(num1: int, num2: int) -> int:
    return num1 + num2
```

```python
print(sum_two_numbers(1, 2))
```

## Lambda

```python
multiple_two_numbers = lambda num1, num2: num1 * num2
```

```python
print(multiple_two_numbers(2, 2))
```

```python
result = (lambda num1, num2: num1 - num2)(3, 2)

print(result)
```

# Arguments

## Args and kwargs

```python
def args_and_kwargs(param1, *args, **kwargs):
    print(f"{param1} args", end="\n")
    for arg in args:
        print(arg)
    
    print(f"{param1} kwargs")
    for kwarg in kwargs:
        print(kwarg, kwargs[kwarg])
```

```python
args_and_kwargs("These are", 1, 2, 3, four=4, five=5, six=6)
```

### Args

```python
def get_max(*params):
    return max(params)
```

```python
numbers = [1, 2, 3]

get_max(*numbers)
```

### Kwargs

```python
def get_greetings(title: str, name: str) -> str:
    return f"{title}, {name}"
```

```python
params = dict(title="Hello", name="world")

get_greetings(**params)
```

# Generators

```python
def fibonacci_sequence(n: int):
    a, b = 0, 1

    for _ in range(n):
        yield a
        a, b = b, a + b
```

```python
for i in fibonacci_sequence(10):
    print(i)
```
