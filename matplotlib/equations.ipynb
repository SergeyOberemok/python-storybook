```python
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import display, Math
```

```python
from sympy.abc import x, y
```

---

# Equation

```python
exp = x**2 - 3*x - y

display(Math(sp.latex(exp)))
```

```python
from_range = -11
to_range = 11

exp = x**2 - 3*x - y

for i in range(from_range, to_range):
    result = next(iter(sp.solve(exp.subs({ 'x': i }), y)))
    plt.plot(i, result, 'o')

plt.grid()

plt.show()
```

---

```python
exp = abs(x) ** (1/2) - y

display(Math(sp.latex(exp)))
display(Math(sp.latex(next(iter(sp.solve(exp, y))))))
```

```python
max_number = 20

for i in range(-max_number, max_number + 1):
    j = abs(i) ** (1/2)
    plt.plot([0, i], [0, j])

plt.show()
```
