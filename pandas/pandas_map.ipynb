```python
import pandas as pd
from faker import Faker
```

```python
fake = Faker()
```

---

```python
df = pd.DataFrame({
    'name': [fake.first_name() for _ in range(3)],
    'age': [fake.random_number(digits=2) for _ in range(3)]
})
```

```python
df.head()
```

### apply - DataFrame lever

##### row axis

```python
series = df.apply(lambda row: row['age'] / 10, axis=1)
series.to_frame(name='age')
```

##### column axis

```python
df.apply(lambda column: column.sum(), axis=0)
```

### transform - column level

```python
df['age'].transform(lambda age: age / 10)
```

### Map

```python
df['age'].map(lambda age: age / 10)
```
