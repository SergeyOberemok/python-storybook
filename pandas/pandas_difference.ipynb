```python
import pandas as pd
from faker import Faker
import numpy as np
```

```python
fake = Faker()
```

---

```python
arr1 = [fake.first_name() for _ in range(3)]
arr2 = [*arr1, fake.first_name()]
arr3 = [fake.first_name() for _ in range(3)]

arr1, arr2, arr3
```

```python
s1 = pd.Series(arr1)
s2 = pd.Series(arr2)
s3 = pd.Series(arr3)
```

---

### Equality

```python
s1.compare(s3) # should be with the same length
```

### Duplicates | Intersect

```python
s1[s1.isin(s2)]
```

```python
np.intersect1d(s1, s2)
```

### Unique

```python
s2[~s2.isin(s1)]
```
