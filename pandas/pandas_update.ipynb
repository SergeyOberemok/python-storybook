CrossTab

```python
data = {
    "Gender": ["M", "M", "F", "F", "M", "F", "M"],
    "AgeGroup": ["18-30", "18-30", "18-30", "31-50", "31-50", "31-50", "18-30"],
    "Purchased": [1, 0, 1, 1, 0, 1, 1],
}
```

```python
cross_tab_df = pd.DataFrame(data)
```

```python
cross_tab = pd.crosstab(index=cross_tab_df["Gender"], columns=cross_tab_df["AgeGroup"], values=cross_tab_df["Purchased"], aggfunc="sum")
```

```python
print(cross_tab)
```
